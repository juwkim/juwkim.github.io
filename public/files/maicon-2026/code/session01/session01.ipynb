### 정리

```python
import pandas as pd

# 1. CSV 파일 로드
df = pd.read_csv("maintenance_log.csv")

# 2. 결측치 처리
# repair_count와 repair_duration은 평균값으로 대체
df['repair_count'] = df['repair_count'].fillna(df['repair_count'].mean())
df['repair_duration'] = df['repair_duration'].fillna(df['repair_duration'].mean())

# last_check_day는 중앙값으로 대체
df['last_check_day'] = df['last_check_day'].fillna(df['last_check_day'].median())

# 3. 처리된 결과 저장
df.to_csv("maintenance_cleaned.csv", index=False)
```

# 문제2

```python
import pandas as pd

df = pd.read_csv("health_check.csv")
```

```python
df.describe()
```

```python
# Z-score 직접 계산: Z = (x - mean) / std
df['bmi_z'] = (df['bmi'] - df['bmi'].mean()) / df['bmi'].std()
df['bp_z'] = (df['blood_pressure'] - df['blood_pressure'].mean()) / df['blood_pressure'].std()
```

```python
from scipy.stats import zscore

# Z-score 계산
df[['bmi_z', 'bp_z']] = df[['bmi', 'blood_pressure']].apply(zscore)
```

```python
condition = (df['bmi_z'].abs() <= 3) & (df['bp_z'].abs() <= 3)
cleaned_df = df[condition].drop(columns=['bmi_z', 'bp_z'])
```

```python
num_removed = len(df) - len(cleaned_df)
print(f"제거된 이상치 행 수: {num_removed}")
```

```python
cleaned_df.to_csv("health_clean.csv", index=False)
```

### 정리

```python
import pandas as pd
from scipy.stats import zscore

# 1. CSV 파일 로드
df = pd.read_csv("health_check.csv")

# 2. Z-score 계산
# df['bmi_z'] = (df['bmi'] - df['bmi'].mean()) / df['bmi'].std()
# df['bp_z'] = (df['blood_pressure'] - df['blood_pressure'].mean()) / df['blood_pressure'].std()
df[['bmi_z', 'bp_z']] = df[['bmi', 'blood_pressure']].apply(zscore)

# 3. Z-score가 3을 초과하는 이상치 제거
condition = (df['bmi_z'].abs() <= 3) & (df['bp_z'].abs() <= 3)
cleaned_df = df[condition].drop(columns=['bmi_z', 'bp_z'])

# 4. 제거된 행 수 출력
num_removed = len(df) - len(cleaned_df)
print(f"제거된 이상치 행 수: {num_removed}")

# 5. 정제된 데이터 저장
cleaned_df.to_csv("health_clean.csv", index=False)
```

### 정리

```python
import pandas as pd

# 1. 센서 데이터 로드
motion_df = pd.read_csv("motion_sensor.csv")
temp_df = pd.read_csv("temp_sensor.csv")

# 2. 두 센서 데이터를 시간 기준으로 병합
merged_df = pd.merge(motion_df, temp_df, on="time", suffixes=('_motion', '_temp'))

# 3. motion_count에서 'error' → NaN으로 처리
merged_df['motion_count'] = pd.to_numeric(merged_df['motion_count'], errors='coerce')

# 4. 평균으로 결측치 대체 (motion_count만 해당)
motion_mean = merged_df['motion_count'].mean()
merged_df['motion_count'] = merged_df['motion_count'].fillna(motion_mean)

# 5. 결측치가 남아 있는 행 제거
cleaned_df = merged_df.dropna()

# 6. 결과 저장
cleaned_df.to_csv("merged_sensor_cleaned.csv", index=False)
```
