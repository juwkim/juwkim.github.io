```python
import pandas as pd

df = pd.read_csv("stress_analysis.csv")
```

```python
df.info()
```

```python
correlation = df["sleep_quality"].corr(df["training_pressure"])
```

```python
correlation # -0.94
```

[해석] 훈련 강도가 높으면 수면의 질이 떨어진다. (수면의 질과 훈련 강도는 강한 반비례 관계를 가진다.)

```python
threshold = df["command_tension"].median()
```

```python
threshold
```

```python
high_group_mean = df[df["command_tension"] > threshold]["sleep_quality"].mean()
low_group_mean = df[df["command_tension"] <= threshold]["sleep_quality"].mean()
```

```python
high_group_mean, low_group_mean # 2.63, 6.83
```

[해석] 명령의 강도(긴장도)가 낮을수록 수면의 질이 비교적 상승한다. (명령의 강도(긴장도)가 높은 그룹은 수면의 질이 2.63 정도로 낮고, 낮은 그룹은 6.83 정도로 높다.)

```python
summary = {
    "sleep-training_correlation": round(correlation, 2),
    "high_command_tension_sleep_avg": round(high_group_mean, 2),
    "low_command_tension_sleep_avg": round(low_group_mean, 2),
    "command_tension_threshold": threshold
}
```

```python
summary
```

### 정리

```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("stress_analysis.csv")

# 1. sleep_quality와 training_pressure 간의 상관관계 분석
correlation = df["sleep_quality"].corr(df["training_pressure"])

# 2. command_tension 기준으로 높은 그룹과 낮은 그룹 나눠 평균 비교
threshold = df["command_tension"].median()
high_group_mean = df[df["command_tension"] > threshold]["sleep_quality"].mean()
low_group_mean = df[df["command_tension"] <= threshold]["sleep_quality"].mean()

# 결과 요약
summary = {
    "sleep-training_correlation": round(correlation, 2),
    "high_command_tension_sleep_avg": round(high_group_mean, 2),
    "low_command_tension_sleep_avg": round(low_group_mean, 2),
    "command_tension_threshold": threshold
}

print(summary)
```

### 정리

```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("meal_feedback.csv")

# 1. 부대별 평균 만족도 계산
unit_avg = df.groupby("unit")[["taste_score", "quantity_score", "cleanliness_score"]].mean()

# 2. 평균이 가장 낮은 항목명 찾기 (전체 평균 기준)
col_means = df[["taste_score", "quantity_score", "cleanliness_score"]].mean()
lowest_avg_column = col_means.idxmin()

# 3. 저장
output_path = "meal_unit_avg.csv"
unit_avg.to_csv(output_path)

# 4. pivot_table 사용 → 구조화
pivot_avg = pd.pivot_table(df, index="unit", values=["taste_score", "quantity_score", "cleanliness_score"], aggfunc="mean")
```

### 정리

```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("equipment_operational.csv")

# 1. 장비 유형별 연료 효율 및 수리 빈도 평균 계산
type_avg = df.groupby("type")[["fuel_efficiency", "repair_freq"]].mean()

# 2. 수리 빈도 기준으로 효율이 가장 높은 장비 유형 판단 (낮을수록 우수)
most_efficient_type = type_avg.sort_values("repair_freq").index[0]

# 3. 저장
output_path = "equipment_type_avg.csv"
type_avg.to_csv(output_path)

# 4. pivot_table로 동일 요약 수행
pivot_avg = pd.pivot_table(df, index="type", values=["fuel_efficiency", "repair_freq"], aggfunc="mean")
```
