### 정리

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. CSV 파일 로드
df = pd.read_csv("sensor_log.csv")

# 2. timestamp를 datetime 형식으로 변환
df['timestamp'] = pd.to_datetime(df['timestamp'])

# 3. timestamp 기준 정렬
df = df.sort_values('timestamp')

# 4. 시간 인덱스로 설정
df.set_index('timestamp', inplace=True)

# 5. 1시간 단위로 motion_detected 합계 계산
hourly_sum = df['motion_detected'].resample('1H').sum()

# 6. 결과 저장
hourly_sum.to_csv("hourly_motion.csv", header=["motion_sum"])

# 7. 시계열 그래프 시각화
plt.figure(figsize=(10, 4))
hourly_sum.plot(marker='o')
plt.title("시간대별 감지된 움직임 합계")
plt.xlabel("시간")
plt.ylabel("motion_sum")
plt.grid(True)
plt.tight_layout()
plt.savefig("hourly_motion_plot.png")
```

### 정리

```python
import pandas as pd
import json

# 1. JSON 파일 로드 및 정규화
with open("supply.json", encoding="utf-8") as f:
    raw_data = json.load(f)

# 2. 중첩 구조 풀기: 'items' 딕셔너리를 컬럼으로 펼침
df = pd.json_normalize(raw_data)

# 3. 컬럼 이름 정리
df = df.rename(columns={
    'items.water': 'water',
    'items.ration': 'ration',
    'items.medicine': 'medicine'
})

# 4. CSV 저장
csv_path = "supply_log.csv"
df.to_csv(csv_path, index=False)

# 5. 항목별 총합 계산 및 출력
totals = df[['water', 'ration', 'medicine']].sum()
print(totals)
```
