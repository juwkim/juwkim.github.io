### 정리

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 1. CSV 파일 로드
df = pd.read_csv("fitness_test.csv")

# 2. 정규화할 컬럼 지정
features = ['pushup_count', 'run_time_2km', 'situp_count']

# 3. MinMaxScaler 적용
scaler = MinMaxScaler()
scaled_values = scaler.fit_transform(df[features])

# 4. 정규화된 결과를 새로운 컬럼으로 추가
scaled_df = pd.DataFrame(scaled_values, columns=[f"{col}_scaled" for col in features])
df_scaled = pd.concat([df, scaled_df], axis=1)

# 5. 결과 저장
df_scaled.to_csv("fitness_scaled.csv", index=False)

# 6. 각 원본 컬럼의 최소/최대값 출력
for col in features:
    print(f"{col} 최소값: {df[col].min()}, 최대값: {df[col].max()}")
```

### 정리

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. 데이터 로드
df = pd.read_csv("soldier_info.csv")

# 2. Label Encoding: 'position'
le = LabelEncoder()
df['position_encoded'] = le.fit_transform(df['position'])

# 🔎 position_encoded
# 정비병 → 0, 통신병 → 1, 보병 → 2, 의무병 → 3, 포병 → 4 (예시, 실제 매핑은 le.classes_ 확인)

# 3. One-Hot Encoding: 'region'
region_encoded = pd.get_dummies(df['region'], prefix='region')

# 4. 인코딩 결과 병합
encoded_df = pd.concat([df, region_encoded], axis=1)

# 5. 결과 저장
encoded_df.to_csv("encoded_soldiers.csv", index=False, encoding='utf-8-sig')

# 6. 인코딩 컬럼 설명 출력
print("# position (Label Encoding):", dict(zip(le.classes_, le.transform(le.classes_))))
print("# region (One-Hot Encoding):", list(region_encoded.columns))
```

```python
import pandas as pd

df = pd.read_csv("video_metadata.csv")
```

```python
df.head()
```

```python
import os

df['exists'] = df['file_path'].apply(lambda path: os.path.exists(path))
```

```python
valid_df = df[df['exists'] == True]
missing_df = df[df['exists'] == False]
```

```python
valid_df.to_csv("valid_videos.csv", index=False)
missing_df.to_csv("missing_videos.csv", index=False)
```

```python
print(f"존재하는 파일 수: {len(valid_df)}")
print(f"누락된 파일 수: {len(missing_df)}")
```

### 정리

```python
import pandas as pd
import os

# 1. CSV 파일 로드
df = pd.read_csv("video_metadata.csv")

# 2. file_path 존재 여부 확인
df['exists'] = df['file_path'].apply(lambda path: os.path.exists(path))

# 3. 존재하는 파일과 누락된 파일 분리
valid_df = df[df['exists'] == True]
missing_df = df[df['exists'] == False]

# 4. 파일 저장
valid_df.to_csv("valid_videos.csv", index=False)
missing_df.to_csv("missing_videos.csv", index=False)
```
