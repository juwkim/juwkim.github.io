### 정리

```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 로드
fitness_df = pd.read_csv('fitness_result.csv')

# plots 폴더 경로 생성
output_dir = "plots"
os.makedirs(output_dir, exist_ok=True)

# 시각화 대상 컬럼
columns = ["pushup_count", "run_time_2km(sec)", "situp_count"]

# 시각화 및 저장
for col in columns:
    plt.figure(figsize=(10, 4))

    # 1-1. 히스토그램
    plt.subplot(1, 2, 1)
    plt.hist(fitness_df[col], bins=15, color='skyblue', edgecolor='black')
    plt.title(f'{col} Histogram')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    # 1-2. 박스플롯
    plt.subplot(1, 2, 2)
    sns.boxplot(x=fitness_df[col], color='lightgreen')
    plt.title(f'{col} Boxplot')

    # 3. 저장
    filename = os.path.join(output_dir, f"{col}_plot.png")
    plt.tight_layout()
    plt.savefig(filename)
    plt.close()
```

### 정리

```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 로드
stress_df = pd.read_csv('stress_factors.csv')

# 시각화 대상 폴더 생성
output_dir = "plots_stress"
os.makedirs(output_dir, exist_ok=True)

# 1. 산점도: training_pressure vs command_tension
plt.figure(figsize=(7, 5))
sns.scatterplot(data=stress_df, x="training_pressure", y="command_tension", hue="unit")
plt.title("Training Pressure vs Command Tension")
plt.xlabel("Training Pressure")
plt.ylabel("Command Tension")
plt.legend(title="Unit", bbox_to_anchor=(1.05, 1), loc='upper left')
scatter_path = os.path.join(output_dir, "scatter_plot.png")
plt.tight_layout()
plt.savefig(scatter_path, bbox_inches='tight')
plt.close()

# 2. 상관계수 행렬 & 히트맵
corr_matrix = stress_df[["sleep_quality", "training_pressure", "command_tension", "social_support"]].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Stress Factors")
heatmap_path = os.path.join(output_dir, "stress_correlation.png")
plt.savefig(heatmap_path)
plt.close()
```

### 정리

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터 로드
df = pd.read_csv("equipment_performance.csv")

# 1. 수치형 컬럼 선택 및 상관계수 계산
numeric_cols = ["fuel_efficiency", "operation_hours", "repair_freq", "mission_success_rate"]
corr_matrix = df[numeric_cols].corr()

# 2. Heatmap 시각화
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Equipment Performance")

# 3. 연료 효율과 작전 성공률의 상관계수 추출
fuel_success_corr = corr_matrix.loc["fuel_efficiency", "mission_success_rate"]

# 4. 이미지 저장
heatmap_path = "equipment_corr.png"
plt.savefig(heatmap_path, bbox_inches='tight')
plt.close()
```
