### 정리

```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# 데이터 로드
df = pd.read_csv('mental_health.csv')

# 클러스터링 대상 컬럼 선택
features = df[["sleep_quality", "stress_level", "social_connection", "self_confidence"]]

# 1. KMeans 군집화
kmeans = KMeans(n_clusters=3, random_state=42)
df["cluster"] = kmeans.fit_predict(features)

# 2. 클러스터 결과 저장
cluster_result_path = "cluster_result.csv"
df.to_csv(cluster_result_path, index=False)

# 3. 군집별 병사 수 시각화
plt.figure(figsize=(6, 4))
df["cluster"].value_counts().sort_index().plot(kind="bar")
plt.title("병사 군집별 인원 수")
plt.xlabel("Cluster")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plot_path = "cluster_barplot.png"
plt.savefig(plot_path)
plt.close()

# 4. 클러스터별 평균 특성 계산
cluster_means = df.groupby("cluster")[["sleep_quality", "stress_level", "social_connection", "self_confidence"]].mean()
print(cluster_means)
```

- PC2는 주로 작전 거리와 전력 소모량의 대조적 특성에 기반 (장거리 저전력/단거리 고전력)

### 정리

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터 로드
df_equipment = pd.read_csv('equipment_features.csv')

# 장비 ID를 제외한 특성 선택
features_eq = df_equipment.drop(columns=["equipment_id"])

# 1. 정규화
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features_eq)

# 1. PCA 수행
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_features)

# 2. 결과 데이터프레임 생성 및 저장
pca_df = pd.DataFrame(pca_result, columns=["PC1", "PC2"])
pca_df["equipment_id"] = df_equipment["equipment_id"]

pca_csv_path = "pca_result.csv"
pca_df.to_csv(pca_csv_path, index=False)

# 3. 시각화
plt.figure(figsize=(8, 6))
sns.scatterplot(data=pca_df, x="PC1", y="PC2")
plt.title("장비 특성의 2차원 PCA 시각화")
plt.xlabel("주성분 1 (PC1)")
plt.ylabel("주성분 2 (PC2)")
plt.tight_layout()
pca_plot_path = "pca_2d.png"
plt.savefig(pca_plot_path)
plt.close()

# 4. 주성분 해석 (주성분 성분)
pca_components = pd.DataFrame(pca.components_, columns=features_eq.columns, index=["PC1", "PC2"])
print(pca_components)
```

```python
import pandas as pd

df_duty = pd.read_csv('duty_pattern.csv')
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_duty)
df_duty["PC1"] = pca_result[:, 0]
df_duty["PC2"] = pca_result[:, 1]
```

```python
df_duty
```

```python
cluster_summary = df_duty.groupby("cluster")[["avg_night_shift_per_week", "avg_patrol_hours", "rest_time", "alert_incidents"]].mean()
```

### 정리

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터 로드 및 전처리
df_duty = pd.read_csv('duty_pattern.csv')
features_duty = df_duty.drop(columns=["soldier_id"])

# 1. 표준화
scaler = StandardScaler()
scaled_duty = scaler.fit_transform(features_duty)

# 1. KMeans 군집화
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(scaled_duty)
df_duty["cluster"] = clusters

# 2. PCA로 2차원 축소
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_duty)
df_duty["PC1"] = pca_result[:, 0]
df_duty["PC2"] = pca_result[:, 1]

# 3. 군집별 평균 특성 요약
cluster_summary = df_duty.groupby("cluster")[["avg_night_shift_per_week", "avg_patrol_hours", "rest_time", "alert_incidents"]].mean()

# 4. 결과 저장
duty_cluster_csv_path = "duty_clusters.csv"
df_duty.to_csv(duty_cluster_csv_path, index=False)

# 4. 시각화
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_duty, x="PC1", y="PC2", hue="cluster", palette="Set2")
plt.title("병사 근무 패턴의 KMeans + PCA 시각화")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend(title="Cluster")
plt.tight_layout()
duty_pca_plot_path = "duty_pca_plot.png"
plt.savefig(duty_pca_plot_path)
plt.close()
```
